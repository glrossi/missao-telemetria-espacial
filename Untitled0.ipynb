# Sistema de Validação de Telemetria e Análise Energética Espacial

Este notebook contém o script de validação de parâmetros de telemetria pré-lançamento e os cálculos de autonomia energética da espaçonave.

```python
# Script de Validação de Telemetria de Decolagem

def verificar_decolagem(telemetria):
    alertas = []

    # 1. Temperatura Interna (18 a 26 °C)
    if not (18 <= telemetria["temp_interna"] <= 26):
        alertas.append(f"Temperatura Interna fora da faixa: {telemetria['temp_interna']}°C")

    # 2. Temperatura Externa (-50 a 110 °C)
    if not (-50 <= telemetria["temp_externa"] <= 110):
        alertas.append(f"Temperatura Externa fora da faixa: {telemetria['temp_externa']}°C")

    # 3. Integridade Estrutural (1 = OK, 0 = Falha)
    if telemetria["integridade_estrutural"] != 1:
        alertas.append("Falha na Integridade Estrutural detectada!")

    # 4. Nível de Energia (>= 80%)
    if telemetria["nivel_energia"] < 80:
        alertas.append(f"Nível de energia insuficiente: {telemetria['nivel_energia']}%")

    # 5. Pressão dos Tanques (100 a 300 psi)
    if not (100 <= telemetria["pressao_tanques"] <= 300):
        alertas.append(f"Pressão dos tanques fora da faixa: {telemetria['pressao_tanques']} psi")

    # 6. Módulos Críticos
    if telemetria["status_modulos"] != "OK":
        alertas.append("Falha identificada nos Módulos Críticos!")

    # Exibição do Resultado
    print("=" * 45)
    print("      RELATÓRIO DE CHECAGEM PRÉ-VOO      ")
    print("=" * 45)
    if not alertas:
        print("STATUS FINAL: >>> PRONTO PARA DECOLAR <<<")
    else:
        print("STATUS FINAL: >>> DECOLAGEM ABORTADA <<<")
        print("\nMotivos do Aborto:")
        for alerta in alertas:
            print(f"- {alerta}")
    print("=" * 45)

# Simulação de Teste
dados_telemetria = {
    "temp_interna": 22.5,
    "temp_externa": 45.0,
    "integridade_estrutural": 1,
    "nivel_energia": 85.0,
    "pressao_tanques": 210.0,
    "status_modulos": "OK"
}

verificar_decolagem(dados_telemetria)
```
